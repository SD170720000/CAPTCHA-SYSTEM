```python
import pandas as pd
```

```python
original_data = pd.read_csv('scenes.csv')
original_data
```

```python
# Give me count of True False in answer column
original_data['answer'].value_counts()
```

```python
chatgpt_predictions = pd.read_csv('chatgpt_predictions.csv')
chatgpt_predictions
```

```python
# Compare the original answers with ChatGPT's predictions

# Take column answer column from both the dataframes
original_answers = original_data['answer']
predicted_answers = chatgpt_predictions['answer']
comparison = original_answers == predicted_answers
accuracy = comparison.mean()
print(f'ChatGPT Accuracy: {accuracy * 100:.2f}%')
```
